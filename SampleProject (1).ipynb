```python
"""
online shopping cart system
This program allows customers to browse, add, remove, and purchase products.
"""


customers = {}
available_products = [
    {"code": "P001", "name": "Laptop", "price": 2999.99, "category": "electronics"},
    {"code": "P002", "name": "Headphones", "price": 149.99, "category": "electronics"},
    {"code": "P003", "name": "T-shirt", "price": 49.99, "category": "clothing"},
    {"code": "P004", "name": "coffee maker", "price": 199.99, "category": "home"},
    {"code": "P005", "name": "python book", "price": 89.99, "category": "books"},
]
print(available_products)
```

```python
from os import name


def add_customer(customer_id, name, email):
    """Add a new customer to the system."""
    customers[customer_id] = {
        "name": name.title(),
        "email": email.lower(),
        "cart": set(),
    }
print(f"✅ Customer '{ name.title()}' added successfully.")
```

```python
def view_products(category=None):
    """Display all available products, optionally filtered by category."""
    print("\n📦 Available Products:")
    print(f"{'Code':<6} {'Name':<15} {'Price':>10}  Category")
    print("-" * 45)
    for p in available_products:
        if category is None or p["category"].lower() == category .lower():
            print(f"{p['code']:<6}  {p['name']:<15} ${p['price']:>9.2f}    {p['category']}")
    print()
```

```python
def search_products(keyword):
    """Search for a product by name or code."""
    keyword = keyword.lower()
    found = False
    for p in available_products:
        if keyword in p["code"].lower() or keyword in p["name"].lower():
            print(f"🔍 Found: {p['code']} - {p['name']} (${p['price']:.2f}) ")
            found = True
    if not found:
        print("❌ No products found.")
```

```python
def add_to_cart(customer_id, product_code):
    """Add a product to the customer's cart (Set prevents duplicates)."""
    product_codes = [p["code"] for p in available_products]

    if customer_id not in customers:
        print("❌ Customer not found.")
    elif product_code not in product_codes:
        print("​❌​ Product not available.")
    elif product_code in customers[customer_id]["cart"]:
        print("⚠️ Product already in cart.")
    else:
        customers[customer_id]["cart"].add(product_code)
        print(f"​🛒 Product '{product_code}' added to cart successfully.")
```

```python
def remove_from_cart(customer_id, product_code):
    """Remove a product from the customer's cart."""
    if product_code in customers[customer_id]["cart"]:
        customers[customer_id]["cart"].remove(product_code)
        print(f"🗑️​ Product '{product_code}' removed from cart.")
    else:
        print("❌ Product not found in cart.")
```

```python
def view_cart(customer_id):
    """Display the customer's current cart ."""
    if customer_id not in customers:
        print("❌ Customer not found.")
        return 
    cart = customers[customer_id]["cart"]
    if not cart:
        print("​🛒 cart is empty.")
        return
    print(f"\n🛒 cart for {customers[customer_id]['name']}:")
    print(f"{'Code':<6} {'Name':<15} {'price':>10}")
    print("_" * 35)
    for code in cart:
        for p in available_products:
            if p["code"] == code:
                print(f"{p['code']:<6} {p['name']:<15} ${p['price']:>9.2f}")
    print()
```

```python
"""
Online Shopping Cart System
This program allows customers to browse, add, remove, and checkout products.
"""


customers = {}
available_products = [
    {"code": "P001", "name": "Laptop", "price": 2999.99, "category": "electronics"},
    {"code": "P002", "name": "Headphones", "price": 149.99, "category": "electronics"},
    {"code": "P003", "name": "T-shirt", "price": 49.99, "category": "clothing"},
    {"code": "P004", "name": "coffee maker", "price": 199.99, "category": "home"},
    {"code": "P005", "name": "python book", "price": 89.99, "category": "books"},
]


def add_customer(customer_id, name, email):
    customers[customer_id] = {
        "name": name.title(),
        "email": email.lower(),
        "cart": set(),
    }
    print(f"Customer '{name.title()}' added successfully.")


def view_products():
    print("\n📦 Available Products:")
    print(f"{'Code':<6} {'Name':<15} {'Price':>10}  Category")
    print("-" * 45)
    for p in available_products:
            print(f"{p['code']:<6}  {p['name']:<15} ${p['price']:>9.2f}   {p['category']}")
    print()

def search_products(keyword):
    keyword = keyword.lower()
    found = False
    for p in available_products:
        if keyword in p["code"].lower() or keyword in p["name"].lower():
            print(f"🔍 Found: {p['code']} - {p['name']} (${p['price']:.2f}) ")
            found = True
    if not found:
        print("❌ No products found.")
        
def add_to_cart(customer_id, product_code):
    """Add a product to the customer's cart (Set prevents duplicates)."""
    product_codes = [p["code"] for p in available_products]

    if customer_id not in customers:
        print("❌ Customer not found.")
    elif product_code not in product_codes:
        print("​❌​ Product not available.")
    elif product_code in customers[customer_id]["cart"]:
        print("⚠️ Product already in cart.")
    else:
        customers[customer_id]["cart"].add(product_code)
        print(f"​🛒 Product '{product_code}' added to cart successfully.")

def remove_from_cart(customer_id, product_code):
    """Remove a product from the customer's cart."""
    if product_code in customers[customer_id]["cart"]:
        customers[customer_id]["cart"].remove(product_code)
        print(f"🗑️​ Product '{product_code}' removed from cart.")
    else:
        print("❌ Product not found in cart.")
        
def view_cart(customer_id):
    """Display the customer's current cart ."""
    if customer_id not in customers:
        print("❌ Customer not found.")
        return 
    cart = customers[customer_id]["cart"]
    if not cart:
        print("​🛒 cart is empty.")
        return
    print(f"\n🛒 cart for {customers[customer_id]['name']}:")
    print(f"{'Code':<6} {'Name':<15} {'price':>10}")
    print("_" * 35)
    for code in cart:
        for p in available_products:
            if p["code"] == code:
                print(f"{p['code']:<6} {p['name']:<15} ${p['price']:>9.2f}")
    print()

def checkout_summary(customer_id):
    customer = customers.get(customer_id)
    if not customer:
        return("Not found", 0, 0.0, "Invalid")
    cart = customer["cart"]
    total_items = len(cart)
    total_price = sum(p["price"] for p in available_products if p ["code"] in cart)
    status = "Ready to checkout" if total_items > 0 else "Cart is empty"
    return (customer["name"], total_items, round(total_price, 2), status)

# Demo run
print("=" * 50)
print("   🛍️  Online Shopping Cart System")
print("=" * 50)



add_customer("C001", "Hoor Ahmed", "hoor@email.com")
view_products()
search_products("laptop")
add_to_cart("C001", "P001")
add_to_cart("C001", "P002")
add_to_cart("C001", "P001")    # duplicate - should warn 
remove_from_cart("C001", "P002")
view_cart("C001")
name, items, total, status = checkout_summary("C001")
print(f"📋​ summary  → Customer: {name} ❘ Items: {items} ❘ Total: ${total:.2f} ❘ Status: {status}")
```
